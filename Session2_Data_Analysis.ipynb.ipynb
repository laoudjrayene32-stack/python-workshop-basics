```python

import pandas as pd

fruits = pd.Series({
    "Apple": 120,
    "Banana": 80,
    "Orange": 100,
    "Mango": 200,
    "Grape": 150
})

print(fruits)
```

```python

third_price = fruits.iloc[2]
print("Third fruit price:", third_price)
```

```python

new_prices = fruits * 1.1
print(new_prices)
```

```python

average_price = fruits.mean()
most_expensive = fruits.idxmax()
highest_price = fruits.max()

print("Average price:", average_price)
print("Most expensive fruit:", most_expensive)
print("Highest price:", highest_price)
```

```python

data = {
    "Name": ["Ali", "Sara", "Yasmine", "Omar", "Lina"],
    "Math": [90, 85, 75, 95, 88],
    "English": [80, 92, 78, 87, 90],
    "Science": [85, 89, 82, 91, 86]
}

df = pd.DataFrame(data)
print(df)
```

```python

print(df.head(3))
```

```python

math_scores = df["Math"]
print(math_scores)
```

```python

result = df[df["Math"] > 80]
print(result)
```

```python

result = df[
    (df["Math"] > 85) &
    (df["English"] > 85)
]

print(result)
```

```python

sales_data = {
    "Product": ["A", "A", "B", "B", "C", "C", "A", "B"],
    "Region": [
        "North", "South", "North", "East",
        "West", "South", "West", "South"
    ],
    "Sales": [100, 150, 200, 180, 120, 160, 130, 220]
}

sales_df = pd.DataFrame(sales_data)
print(sales_df)
```

```python

total_sales = sales_df.groupby("Product")["Sales"].sum()
print(total_sales)
```

```python

average_sales = sales_df.groupby("Region")["Sales"].mean()
print(average_sales)
```

```python

result = sales_df.groupby("Product")["Sales"].agg(
    ["mean", "max"]
)

print(result)
```

```python

import matplotlib.pyplot as plt

revenue_data = {
    "Quarter": ["Q1", "Q2", "Q3", "Q4"],
    "Revenue": [1000, 1500, 1300, 2000]
}

revenue_df = pd.DataFrame(revenue_data)

revenue_df.plot(
    x="Quarter",
    y="Revenue",
    kind="line",
    marker="o"
)

plt.title("Quarterly Revenue")
plt.xlabel("Quarter")
plt.ylabel("Revenue")
plt.show()
```

```python

revenue_df.plot(
    x="Quarter",
    y="Revenue",
    kind="bar",
    figsize=(8, 5)
)

plt.title("Quarterly Revenue Comparison")
plt.xlabel("Quarter")
plt.ylabel("Revenue")
plt.xticks(rotation=0)
plt.show()
```

```python

scores_data = {
    "Student": ["Ali", "Sara", "Omar", "Lina", "Yasmine",
                "Amine", "Nour", "Meriem"],
    "Score": [55, 60, 65, 70, 75, 80, 85, 90]
}

scores_df = pd.DataFrame(scores_data)

scores_df["Score"].plot(
    kind="hist",
    bins=5,
    edgecolor="black"
)

plt.title("Distribution of Student Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

iris_df = pd.read_csv(url)

print(iris_df.head())
print("Dataset shape:", iris_df.shape)
```

```python

virginica_df = iris_df[
    iris_df["species"] == "virginica"
]

print(virginica_df.head(5))
```

```python

min_width = iris_df["petal_width"].min()
max_width = iris_df["petal_width"].max()

print("Minimum petal width:", min_width)
print("Maximum petal width:", max_width)
```

```python

average_sepal_length = iris_df.groupby(
    "species"
)["sepal_length"].mean()

print(average_sepal_length)

highest_species = average_sepal_length.idxmax()
print("Species with highest average sepal length:",
      highest_species)
```

```python

iris_df.plot(
    x="petal_length",
    y="petal_width",
    kind="scatter",
    figsize=(8, 5)
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.show()
```

```python

average_petal_length = iris_df.groupby(
    "species"
)["petal_length"].mean()

average_petal_length.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Petal Length (cm)")
plt.xticks(rotation=0)
plt.show()
```
